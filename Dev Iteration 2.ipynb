```python
def fit_text(text, font, linespace, aspect_ratio):
    # Create a sample object to take measurements from:
    measure_object=textlayout.TextMultiLine((0,0), text, linespace, font)
    measure_object_area = measure_object.width * measure_object.height # this is in pixel-land
    text_lines=text.split("\n")
    longest_charlength=max([len(l) for l in text_lines])
    lines_count=len(text_lines)
    inferred_char_width=measure_object.width/longest_charlength
    inferred_char_height=measure_object.height/lines_count
    ww_char_width=sqrt(measure_object_area * aspect_ratio)/inferred_char_width
    ww_expected_lines=sqrt(measure_object_area/aspect_ratio)/inferred_char_height
    return ww_char_width, ww_expected_lines
    
```

```python
def svg_viewbox(window_x, window_y, window_w, window_h, screen_w, screen_h, content=""):
    return f"""<svg xmlns="http://www.w3.org/2000/svg" viewBox="{window_x} {window_y} {window_w} {window_h}" width="{screen_w}" height="{screen_h}" > {content} </svg>"""
```

```python
import base64
def encode_data_to_base64_uri(font_file_loc):
    with open(font_file_loc, "rb") as font_file:
        file_bytes=font_file.read()
    font_data_b64=base64.b64encode(file_bytes).decode("ascii")
    return f"data:application/x-font-ttf;base64,{font_data_b64}"

def css_font_style(font_file_loc, 
                 font_family_name):

    return f"""@font-face {{
            font-family:'{font_family_name}';
            src:url({encode_data_to_base64_uri(font_file_loc)}) format('truetype');
            }}
    """
```
