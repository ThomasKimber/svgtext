```python
fonts_from_dir=[]
for f in os.listdir(fontsdir):
    f_path=os.path.join(fontsdir, f)
    if os.path.isfile(f_path):
        fonts_from_dir.append(f_path)

svg_font_registry = svgsystem.SVGFontRegistry(fonts_from_dir)
```

```python
s="""Mary had a little lamb
Its fleece as white as snow
And everywhere that Mary went
The lamb was sure to go"""
```

```python
viewbox=str(svgsystem.SVGViewBox(-20,-20,220,120,800,400))
defs_block = f"""<defs><style type="text/css">{style_content}</style></defs>"""

svg_layout = svgsystem.SVGMultiLineText(
                                     s,
                                     1.0,
                                     svg_font_registry.registry['Comfortaa Regular'], 12, 
                                     title="A poem about lambs", 
                                     style_class='comfortaa_default')
svg_transform_group_a = svgsystem.SVGTransformMatrix.from_bounds(
    svg_layout.bounds, 
    (50,50,150,90)
    #svg_layout.bounds
    #(50,50,190,90)
)

svg_transform_group_b = svgsystem.SVGTransformMatrix.from_bounds(
    svg_layout.bounds, 
    (10,10,90,40)
    #svg_layout.bounds
    #(50,50,190,90)
)

svg_transform_group_c = svgsystem.SVGTransformMatrix.from_bounds(
    (0,0,105,60), 
    (100,25,200,65)
    #(30,30,145,80)
    #svg_layout.bounds
    #(50,50,190,90)
)

#panel_test = titled_panel(20,5,105,60,0,15)
panel_test = svgsystem.SVGTitledPanel("002", 105, 60, "PETSANDFISHES", (25,5,25,5), svg_font_registry.registry['Comfortaa Regular'], style_class="comfortaa_default")


tlx, tly, brx, bry = svg_layout.layout.bounds
transformed_bounds_a = [d for c in [svg_transform_group_a._apply(tlx, tly), svg_transform_group_a._apply(brx, bry)] for d in c]
transformed_bounds_b = [d for c in [svg_transform_group_b._apply(tlx, tly), svg_transform_group_b._apply(brx, bry)] for d in c]

test_rect_a = svgsystem.SVGRectangle(transformed_bounds_a[0], transformed_bounds_a[1], transformed_bounds_a[2]-transformed_bounds_a[0], transformed_bounds_a[3]-transformed_bounds_a[1],0,0, style_class="layout_rect")
test_rect_b = svgsystem.SVGRectangle(transformed_bounds_b[0], transformed_bounds_b[1], transformed_bounds_b[2]-transformed_bounds_b[0], transformed_bounds_b[3]-transformed_bounds_b[1],0,0, style_class="layout_rect")
layout_grid = str(svgsystem.SVGDebugLayoutGrid("001", 0,0, 200,100, 
                                          "#000000", 
                                          (0,200,20), 
                                          (0,100,10), 
                                          "axes_font", 
                                          12, 
                                          (0,0,0,0)))
```

```python
HTML(viewbox.replace("%%placeholder%%", 
                     defs_block +\
                     layout_grid +\
                     str(svg_transform_group_a).replace("%%placeholder%%", str(svg_layout)) +\
                     str(svg_transform_group_b).replace("%%placeholder%%", str(svg_layout)) +\
                     str(test_rect_a) +\
                     str(test_rect_b) +\
                     str(svg_transform_group_c).replace("%%placeholder%%", str(panel_test))
                    ))
```

```python
class VisAttribute:
    name : str
    fully_qualified_name : str
    description : str
    domain : str
    is_pk : bool
    is_fk : bool
    show_description : bool
    show_key_indicators : bool
    show_domain : bool

    def __init__(self, **kwargs):
        for k,v in kwargs.items():
            setattr(self, k, v)

class VisEntity:
    identifier : str
    name : str
    fully_qualified_name : str
    description : str
    attribute_list : list[VisAttribute]
    show_description : bool
    show_attributes: bool
    show_top_n_attributes: int

    def __init__(self, **kwargs):
        for k,v in kwargs.items():
            setattr(self, k, v)
```

```python
attribute_data = [
    {"name" : "Identifer",
     "fully_qualified_name" : "Person.Identifier",
     "description" : "An identifier that uniquely identifies a person",
     "domain" : "Identifier",
     "is_pk" : True, 
     "is_fk" : False},
    {"name" : "FirstName",
     "fully_qualified_name" : "Person.FirstName",
     "description" : "The first word of a person's full given name.",
     "domain" : "Name",
     "is_pk" : False, 
     "is_fk" : False},
    {"name" : "FamilyName",
     "fully_qualified_name" : "Person.FamilyName",
     "description" : "The word of a person's full given name that is shared by members of their family.",
     "domain" : "Name",
     "is_pk" : False, 
     "is_fk" : False},
    {"name" : "ColourPreference",
     "fully_qualified_name" : "Person.ColourPreference",
     "description" : "A colour expressed by a person as being their preference or \"favourite\".",
     "domain" : "String",
     "is_pk" : False, 
     "is_fk" : False},
    {"name" : "EmailAddress",
     "fully_qualified_name" : "Person.EmailAddress",
     "description" : "An email address by which a person can be contacted.",
     "domain" : "Email",
     "is_pk" : False, 
     "is_fk" : False},
    {"name" : "URL",
     "fully_qualified_name" : "Person.URL",
     "description" : "A URL referencing web-content operated by a person.",
     "domain" : "String",
     "is_pk" : False, 
     "is_fk" : False},
    {"name" : "AddressIdentifier",
     "fully_qualified_name" : "Person.AddressIdentifier",
     "description" : "Reference to the Local record at which this person resides.",
     "domain" : "Identifier",
     "is_pk" : False, 
     "is_fk" : True},
]
person_attributes=[VisAttribute(**d) for d in attribute_data]
entity_data=[
    {"identifier" : "Person",
    "name" : "Person",
    "fully_qualified_name" : "Person",
    "description" : "An individual human being.",
    "attribute_list" : person_attributes}
]

person_entity=[VisEntity(**d) for d in entity_data][0]
```
